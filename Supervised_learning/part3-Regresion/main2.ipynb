# Regresion types
## Linear Regresioin VS UNlinear Regresion


### Import 
import numpy, pandas, matplotlib and linearRegresion from sklearn.linear_model.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as mplt
from sklearn.linear_model import LinearRegression
```

```python
# Datas: x and, y.
# Model: 2 * x - 1
x = np.array([1, 2, 3, 4])
y = np.array([1, 3, 5, 7])
x.shape
# y.shape
```

### Show chart
we can see the chart and know that what's the chart type.

```python
# Plot
mplt.title("Chart_X")
mplt.xlabel("X")
mplt.ylabel("Y")
mplt.plot(x, y, "*")

mplt.show()
```

as we see, the type of this chart is Linear. because the dots are on a simple
line. pay atention to bellow:

```python
mplt.title("Chart_X")
mplt.xlabel("X")
mplt.ylabel("Y")

mplt.plot(x, y, color="green")

mplt.show()
```

as we see, the datas are on a simple line. now we most create a linear Regeresion. than, we use the linearRegresion func from SKlearn.linear_model.


```python
# Model: linearRegresion
x1 = x.reshape(-1, 1)

model_X = LinearRegression()
model_X.fit(x1, y)
```

#### Model: 
        c * x + i
            c = 2
            x = 5
            i = -1
                c = coef_
                x = predicted X
                i = intercept_
    

```python
"""
    Linear equation: y = m * x + c
    c(.coef_[0]) =2.0 * x = 5 + i(.intercept_) = -1.0
        m = c
        c = i
"""

c = model_X.coef_[0]
i = model_X.intercept_

print(f"C(Coefficient) = {c} and I(Intercept) = {i}")
```

```python
# Plot the regresion chart:

mplt.plot(x1,y, "*")
mplt.plot(x1, model_X.predict(x1))

mplt.show()
```

```python
result_x = model_X.predict([[5]])
result_x[0]
```

## Salary prediction:
    Salary prediction based on years of Experience.
    

```python
years_of_experience = np.array([1, 2, 3, 5, 7, 10, 12, 15, 18, 20, 21, 22, 23, 25])
salaries = np.array([40000, 44000, 48500, 50000, 54000, 60500, 65500, 70000, 78000, 8500, 90500, 120000, 150000, 160000])
years_of_experience = years_of_experience.reshape(-1, 1)
years_of_experience.shape
len(salaries)
```

```python
mplt.plot(years_of_experience, salaries, "*")

mplt.show()
```

```python
from sklearn.tree import DecisionTreeRegressor

model_YS = DecisionTreeRegressor(max_depth=4)
model_YS.fit(years_of_experience, salaries)

y_predict = model_YS.predict(years_of_experience)

mplt.title("Experience based salaries ")
mplt.xlabel("years_of_experience")
mplt.ylabel("salaries(p/y)")
mplt.plot(years_of_experience, salaries, "*")
mplt.plot(years_of_experience, y_predict)

mplt.show()
```
