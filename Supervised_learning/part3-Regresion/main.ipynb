## Liner Regresion

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as mpl_pl
```

### Data
they are my datas or dataSet

```python
# y = 2*x - 1
x = np.array([1, 2, 3, 4]).reshape(-1, 1)
y = np.array([1, 3, 5, 7])

mpl_pl.plot(x, y, "*", color="red")
mpl_pl.title("Regresion type selection")
mpl_pl.xlabel("x")
mpl_pl.ylabel("y")

mpl_pl.show()
```

### Regresion selection
we select a type of regresion by atention to above chart.

```python
# Model: Linear_model> linearRegresion.

from sklearn.linear_model import LinearRegression

mdl_x = LinearRegression()

mdl_x.fit(x, y)
    
```

```python
# Y = m * x + c 

m = mdl_x.coef_[0]
c = mdl_x.intercept_

print(m, c)

```

```python
mpl_pl.plot(x, y, "*")
mpl_pl.plot(x, mdl_x.predict(x))
mpl_pl.show()
```

```python
result_x = mdl_x.predict([[5]])
result_x[0]
```
