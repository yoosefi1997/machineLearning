# Clasification
## Performance_Metrics
we have 4 situation:
                    1 - True-positive(TP)
                    2 - True-Negative(TN)
                    3 - False-Positive(FP)
                    4 - False-Negative(FN)
    if our predict was True about a positive instance, it's TP.
    if our predict was True about a Negative instance, it's TN.
    if our predict about a Positive instance was False, it's FP.
    And if out predict about a negative instance was False, it's FN.
Because, mabe we have a positive instance in our class but, the model
predict that it's a negative instance. however,maybe the model make a misstake.
we use them to create some performance metrics. see bellow please:
important performance metrics are:
                                    1 - acurecy
                                    2 - Recall
                                    3 - precision
                                    



```python
y_True = [1, 1, 0, 0, 1, 0, 0, 0, 1, 1]
y_pred = [1, 1, 0, 0, 1, 0, 1, 0, 1, 1]

```

```python
from sklearn.metrics import accuracy_score

acc = accuracy_score(y_True, y_pred)

acc * 100
```

```python
from sklearn.metrics import precision_score

pr_s = precision_score(y_True, y_pred)

pr_s * 100
```

```python
from sklearn.metrics import recall_score

rc_s = recall_score(y_True, y_pred)

rc_s * 100
```
